### Aim
To implement and compare Random Agent, Q-Learning Agent, and Monte Carlo Tree Search (MCTS) Agent for playing Tic-Tac-Toe, and to analyze their performance using game outcomes and different MCTS simulation budgets.

```
import numpy as np
import matplotlib.pyplot as plt
from mcts_games.environment import TicTacToe
from mcts_games.agents import RandomAgent, QLearningAgent, MCTSAgent, train_qlearning
from mcts_games.tournament import run_tournament
```

### 1. Training Q-Learning Agent (50,000 self-play episodes)

```
ql_agent = QLearningAgent(alpha=0.3, gamma=0.9, epsilon=0.2)
train_qlearning(ql_agent, n_episodes=50000)
ql_agent.training = False
print(f"Q-learning training complete. States stored: {len(ql_agent.Q)}")
```

### 2. Multi-Agent Tournament Evaluation

```
random_agent = RandomAgent()
mcts_agent = MCTSAgent(n_simulations=200)

r_r = run_tournament(random_agent, random_agent, n_games=200, name_a='Random', name_b='Random')
ql_r = run_tournament(ql_agent, random_agent, n_games=200, name_a='QL', name_b='Random')
mc_r = run_tournament(mcts_agent, random_agent, n_games=100, name_a='MCTS', name_b='Random')
mc_ql = run_tournament(mcts_agent, ql_agent, n_games=100, name_a='MCTS', name_b='QL')

print("Tournament Results (Wins / Draws / Losses):")
print(f"Random vs Random: {r_r}")
print(f"QL vs Random:     {ql_r}")
print(f"MCTS vs Random:   {mc_r}")
print(f"MCTS vs QL:       {mc_ql}")
```

### 3. Conclusion & Findings
MCTS outperforms offline Q-learning and random baselines by dynamically simulating future trajectories using the UCB1 selection rule. Higher simulation counts yield near-optimal tactical play.
